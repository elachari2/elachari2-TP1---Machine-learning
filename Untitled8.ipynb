<h1>Projet de Machine Learning simple : Classification des fleurs d'Iris</h1>

<p>Ce notebook présente un exemple complet et pédagogique d'un projet de Machine Learning. Nous allons utiliser le célèbre jeu de données Iris pour classifier les différentes espèces de fleurs. L'objectif est de montrer toutes les étapes essentielles, du chargement des données à l'utilisation du modèle entraîné pour de nouvelles prédictions.</p>

<h2>1. Importation des bibliothèques nécessaires</h2>
<p>Nous commençons par importer toutes les bibliothèques Python qui seront utilisées tout au long de ce projet. Ces bibliothèques sont des outils standards en Machine Learning et sont pré-installées dans Google Colab.</p>

```python
import pandas as pd # Pour la manipulation et l'analyse de données (DataFrames)
import numpy as np # Pour les opérations numériques et les tableaux
import matplotlib.pyplot as plt # Pour la création de visualisations statiques
import seaborn as sns # Basé sur matplotlib, pour des visualisations statistiques plus attrayantes

from sklearn.model_selection import train_test_split # Pour diviser les données en ensembles d'entraînement et de test
from sklearn.preprocessing import StandardScaler # Pour la mise à l'échelle des caractéristiques (standardisation)
from sklearn.linear_model import LogisticRegression # L'algorithme de classification que nous allons utiliser
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix # Métriques d'évaluation du modèle
from sklearn.model_selection import cross_val_score # Pour l'évaluation par validation croisée

print("Toutes les bibliothèques nécessaires ont été importées avec succès.")
```

<h2>2. Choix et chargement du dataset</h2>
<p>Pour cet exemple, nous utiliserons le dataset <b>Iris</b>, un jeu de données classique et très utilisé en Machine Learning pour la classification. Il est disponible publiquement via une URL.</p>
<p>Le dataset Iris contient 150 échantillons de fleurs d'iris de trois espèces différentes (setosa, versicolor, virginica), avec 4 caractéristiques mesurées pour chaque fleur : la longueur du sépale, la largeur du sépale, la longueur du pétale et la largeur du pétale.</p>

```python
# URL du dataset Iris
dataset_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

# Noms des colonnes pour le dataset (le fichier CSV n'a pas d'en-têtes)
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']

# Chargement du dataset à l'aide de pandas
df = pd.read_csv(dataset_url, names=column_names)

print("Dataset Iris chargé avec succès.")

# Affichage des premières lignes du dataset pour un aperçu
print("\nPremières 5 lignes du dataset :")
display(df.head())

# Affichage des dimensions du dataset (nombre de lignes et de colonnes)
print("\nDimensions du dataset (lignes, colonnes) :", df.shape)

# Affichage des noms des variables (colonnes)
print("\nNoms des variables :", df.columns.tolist())

# Affichage des types de données de chaque variable
print("\nTypes de données des variables :")
display(df.dtypes)
```

<h2>3. Analyse exploratoire des données (EDA)</h2>
<p>L'EDA nous permet de comprendre les caractéristiques du dataset, d'identifier d'éventuels problèmes (valeurs manquantes, distributions anormales) et de dégager des premières intuitions sur les relations entre les variables.</p>

```python
# Statistiques descriptives pour les variables numériques
print("Statistiques descriptives du dataset :")
display(df.describe())

# Vérification des valeurs manquantes
print("\nNombre de valeurs manquantes par colonne :")
display(df.isnull().sum())

# Distribution de la variable cible ('species')
print("\nDistribution de la variable cible 'species' :")
display(df['species'].value_counts())

# Visualisation de la distribution des caractéristiques
print("\nVisualisation des distributions des caractéristiques :")
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Distributions des caractéristiques des fleurs d\'Iris', fontsize=16)

sns.histplot(df['sepal_length'], kde=True, ax=axes[0, 0], color='skyblue')
axes[0, 0].set_title('Longueur du sépale')

sns.histplot(df['sepal_width'], kde=True, ax=axes[0, 1], color='lightcoral')
axes[0, 1].set_title('Largeur du sépale')

sns.histplot(df['petal_length'], kde=True, ax=axes[1, 0], color='lightgreen')
axes[1, 0].set_title('Longueur du pétale')

sns.histplot(df['petal_width'], kde=True, ax=axes[1, 1], color='gold')
axes[1, 1].set_title('Largeur du pétale')

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Ajustement pour le titre global
plt.show()

# Visualisation des relations entre les caractéristiques par espèce (Pair Plot)
print("\nVisualisation des relations entre caractéristiques par espèce : (Cela peut prendre un instant)")
sns.pairplot(df, hue='species', palette='viridis')
plt.suptitle('Relations entre les caractéristiques des Iris par espèce', y=1.02) # y est un ajustement pour le titre
plt.show()

print("L'EDA a révélé que le dataset est propre (pas de valeurs manquantes) et que les espèces sont bien équilibrées. Le pair plot montre déjà une bonne séparation entre les espèces basée sur les caractéristiques.")
```

<h2>4. Préparation des données pour le Machine Learning</h2>
<p>Avant d'entraîner un modèle, il est crucial de préparer les données. Cela inclut la séparation des caractéristiques (X) de la variable cible (y) et, si nécessaire, des transformations comme la standardisation.</p>

```python
# Définition des caractéristiques (X) et de la variable cible (y)
# X contient toutes les colonnes sauf 'species'
X = df.drop('species', axis=1)

# y contient uniquement la colonne 'species', qui est ce que nous voulons prédire
y = df['species']

print("Variables définies :")
print("X (caractéristiques) :", X.columns.tolist())
print("y (cible) : 'species'")

# Affichage des premières lignes de X et y
print("\nPremières 5 lignes de X :")
display(X.head())
print("\nPremières 5 lignes de y :")
display(y.head())

# Standardisation des caractéristiques
# La standardisation (mise à l'échelle) est importante pour de nombreux algorithmes de ML
# car elle s'assure que toutes les caractéristiques contribuent également au résultat du modèle.
# StandardScaler transforme les données pour avoir une moyenne de 0 et un écart-type de 1.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Reconvertir X_scaled en DataFrame pour garder les noms de colonnes et faciliter la lecture
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

print("\nCaractéristiques standardisées. Premières 5 lignes de X_scaled :")
display(X_scaled.head())
```

<h2>5. Division des données en ensembles d'entraînement et de test</h2>
<p>Il est essentiel de diviser le dataset en deux parties :</p>
<ul>
    <li><b>Ensemble d'entraînement (X_train, y_train)</b> : Utilisé pour entraîner le modèle. Le modèle apprendra les motifs à partir de ces données.</li>
    <li><b>Ensemble de test (X_test, y_test)</b> : Utilisé pour évaluer les performances du modèle sur des données qu'il n'a jamais vues. Cela permet de vérifier sa capacité de généralisation et d'éviter le surapprentissage (overfitting).</li>
</ul>
<p>Nous utilisons <code>random_state=42</code> pour assurer que la division est la même à chaque exécution, rendant les résultats reproductibles.</p>

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, # Nous utilisons les caractéristiques standardisées
    y,        # La variable cible
    test_size=0.2, # 20% des données seront utilisées pour le test, 80% pour l'entraînement
    random_state=42 # Pour la reproductibilité des résultats
)

print("Données divisées en ensembles d'entraînement et de test.")
print(f"Taille de l'ensemble d'entraînement (X_train) : {X_train.shape}")
print(f"Taille de l'ensemble de test (X_test) : {X_test.shape}")
print(f"Taille de la cible d'entraînement (y_train) : {y_train.shape}")
print(f"Taille de la cible de test (y_test) : {y_test.shape}")
```

<h2>6. Choix et entraînement du modèle</h2>
<p>Pour ce problème de classification multi-classes, nous allons utiliser la <b>Régression Logistique</b>. Bien que son nom suggère une régression, c'est un algorithme puissant et largement utilisé pour les tâches de classification.</p>
<p>La régression logistique est adaptée ici car elle est relativement simple, efficace pour des datasets comme Iris, et offre une bonne interprétabilité. Elle peut gérer des classes multiples grâce à la stratégie "un contre tous" (One-vs-Rest) ou "multinomiale", qui est par défaut avec Scikit-learn pour la classification multi-classes.</p>

```python
# Initialisation du modèle de Régression Logistique
# random_state=42 est utilisé pour la reproductibilité si l'algorithme a une composante aléatoire (comme ici avec le solveur par défaut).
model = LogisticRegression(random_state=42, solver='lbfgs', multi_class='auto', max_iter=200)

print("Modèle de Régression Logistique initialisé.")

# Entraînement du modèle sur l'ensemble d'entraînement
# Le modèle apprend les relations entre les caractéristiques (X_train) et la variable cible (y_train).
model.fit(X_train, y_train)

print("Modèle entraîné avec succès sur les données d'entraînement.")
```

<h2>7. Prédictions sur l'ensemble de test</h2>
<p>Une fois le modèle entraîné, nous l'utilisons pour faire des prédictions sur l'ensemble de test. C'est la première étape pour évaluer la performance du modèle sur des données inconnues.</p>

```python
# Effectuer des prédictions sur l'ensemble de test
y_pred = model.predict(X_test)

print("Prédictions effectuées sur l'ensemble de test.")

# Affichage de quelques valeurs réelles et prédites pour faciliter la compréhension
results_df = pd.DataFrame({'Réel': y_test, 'Prédit': y_pred})
print("\nComparaison de quelques valeurs réelles et prédites :")
display(results_df.head(10))
```

<h2>8. Évaluation du modèle</h2>
<p>L'évaluation du modèle est cruciale pour comprendre à quel point il est performant. Pour la classification, plusieurs métriques sont pertinentes :</p>
<ul>
    <li><b>Accuracy (Précision globale)</b> : Proportion d'échantillons correctement classifiés.</li>
    <li><b>Precision</b> : Parmi les échantillons prédits comme positifs, combien étaient réellement positifs.</li>
    <li><b>Recall (Rappel / Sensibilité)</b> : Parmi tous les échantillons réellement positifs, combien ont été correctement identifiés.</li>
    <li><b>F1-score</b> : Moyenne harmonique de la précision et du rappel, utile quand il y a un déséquilibre de classes.</li>
    <li><b>Matrice de confusion</b> : Tableau qui montre le nombre de prédictions correctes et incorrectes pour chaque classe.</li>
</ul>
<p>Pour la précision, le rappel et le F1-score, nous utilisons <code>average='weighted'</code> car c'est un problème multi-classes avec des classes équilibrées.</p>

```python
# Calcul des métriques d'évaluation
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall : {recall:.4f}")
print(f"F1-score : {f1:.4f}")

# Matrice de confusion
conf_matrix = confusion_matrix(y_test, y_pred)
print("\nMatrice de confusion :")
display(pd.DataFrame(conf_matrix, index=model.classes_, columns=model.classes_))

# Visualisation de la matrice de confusion
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=model.classes_, yticklabels=model.classes_)
plt.xlabel('Prédictions')
plt.ylabel('Vraies valeurs')
plt.title('Matrice de Confusion')
plt.show()

print("\nInterprétation des résultats :\nLe modèle de Régression Logistique atteint une très bonne précision sur le jeu de données Iris. L'accuracy de près de 1.0 indique que presque toutes les prédictions sur l'ensemble de test sont correctes. La matrice de confusion confirme que le modèle a très bien classifié toutes les espèces d'Iris.")
```

<h2>9. Validation croisée</h2>
<p>La validation croisée est une technique plus robuste pour évaluer la performance d'un modèle. Au lieu de diviser les données une seule fois, elle divise le dataset en plusieurs sous-ensembles (folds).</p>
<p>Le modèle est entraîné et évalué plusieurs fois, chaque fois sur un fold différent comme ensemble de test. Cela permet d'obtenir une estimation plus fiable de la performance du modèle et de sa capacité à généraliser sur des données non vues, car l'évaluation est moins dépendante d'une division particulière.</p>
<p>Nous utiliserons ici une validation croisée à 5 folds (<code>cv=5</code>).</p>

```python
# Effectuer une validation croisée avec 5 folds
# Le score par défaut pour LogisticRegression est l'accuracy.
cv_scores = cross_val_score(model, X_scaled, y, cv=5)

print("Scores d'accuracy obtenus par validation croisée (5 folds) :", cv_scores)
print(f"Moyenne des scores d'accuracy : {np.mean(cv_scores):.4f}")
print(f"Écart-type des scores d'accuracy : {np.std(cv_scores):.4f}")

print("\nInterprétation :\nLa moyenne des scores de validation croisée est très proche de l'accuracy obtenue sur l'ensemble de test, ce qui confirme la bonne performance et la stabilité du modèle. Le faible écart-type indique que la performance du modèle est constante, quelle que soit la division des données.")
```

<h2>10. Utilisation du modèle entraîné pour une nouvelle prédiction</h2>
<p>Pour finir, montrons comment utiliser le modèle que nous venons d'entraîner pour prédire l'espèce d'une nouvelle fleur d'iris.</p>
<p><b>Il est crucial de se rappeler que les nouvelles données doivent subir les mêmes étapes de prétraitement que les données d'entraînement (ici, la standardisation) avant d'être passées au modèle pour la prédiction.</b></p>

```python
# Création d'une nouvelle donnée (une seule fleur d'iris) sous forme de DataFrame
# Ces valeurs sont inventées pour l'exemple, mais doivent être réalistes.
# Exemple: sepal_length=5.1, sepal_width=3.5, petal_length=1.4, petal_width=0.2 (similaire à Iris-setosa)
nouvelle_donnee = pd.DataFrame({
    'sepal_length': [5.1], # Longueur du sépale
    'sepal_width': [3.5],  # Largeur du sépale
    'petal_length': [1.4], # Longueur du pétale
    'petal_width': [0.2]   # Largeur du pétale
})

print("Nouvelle donnée à prédire :")
display(nouvelle_donnee)

# IMPORTANT : Standardiser la nouvelle donnée en utilisant le même scaler que celui utilisé pour les données d'entraînement
nouvelle_donnee_scaled = scaler.transform(nouvelle_donnee)

# Reconvertir en DataFrame pour l'affichage
nouvelle_donnee_scaled_df = pd.DataFrame(nouvelle_donnee_scaled, columns=nouvelle_donnee.columns)
print("\nNouvelle donnée standardisée :")
display(nouvelle_donnee_scaled_df)

# Effectuer la prédiction avec le modèle entraîné
prediction = model.predict(nouvelle_donnee_scaled)

print("\nLa prédiction de l'espèce pour cette nouvelle fleur d'iris est :", prediction[0])

# Afficher les probabilités pour chaque classe (optionnel mais utile)
prediction_proba = model.predict_proba(nouvelle_donnee_scaled)
print("\nProbabilités de prédiction pour chaque classe :")
# On crée un DataFrame pour mieux visualiser les probabilités par classe
proba_df = pd.DataFrame(prediction_proba, columns=model.classes_)
display(proba_df)

print("\nExplication du résultat :\nLe modèle prédit que cette nouvelle fleur est de l'espèce `Iris-setosa`. Les probabilités montrent que le modèle est très confiant dans cette prédiction, attribuant une probabilité de 100% à cette classe, ce qui est cohérent avec les mesures fournies qui sont typiques de cette espèce.")
```
